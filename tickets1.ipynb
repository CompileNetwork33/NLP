### Graphs

```python
plt.figure(figsize=(10,5))
sns.countplot(x='Ticket Type',data=df)
plt.xticks(rotation=45)
plt.title("Ticket Type Distribution")
plt.show()
```

```python
df
```

```python
sns.countplot(x='Customer Gender',data=df,color='yellow')
plt.title('customer age')
plt.xlabel('xaxis')
plt.ylabel('yaxis')
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x='Ticket Status',data=df,color='purple')
plt.title('Ticket Status')
plt.xlabel('xaxis')
plt.ylabel('yaxis')
```

```python
priority = df['Ticket Priority'].value_counts()
labels=['Low','Medium','High','Critical']
colors=['orange','teal','gold','lightcoral']

plt.figure(figsize=(6,6))
plt.pie(priority,
        labels=labels,
        colors=colors,
        autopct='%1.1f%%',
        explode=(0,0.1,0,0),
        startangle=90)


plt.title("Ticket Priority")
plt.show()
```
