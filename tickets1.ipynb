### Graphs

```python
plt.figure(figsize=(10,5))
sns.countplot(x='Ticket Type',data=df)
plt.xticks(rotation=45)
plt.title("Ticket Type Distribution")
plt.show()
```

```python
sns.countplot(x='Ticket Channel',data=df,color='yellow')
plt.title('Ticket Channel')
plt.xlabel('xaxis')
plt.ylabel('yaxis')
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x='Ticket Status',data=df,color='purple')
plt.title('Ticket Status')
plt.xlabel('xaxis')
plt.ylabel('yaxis')
```

```python
priority = df['Ticket Priority'].value_counts()
labels=['Low','Medium','High','Critical']
colors=['orange','teal','gold','lightcoral']

plt.figure(figsize=(6,6))
plt.pie(priority,
        labels=labels,
        colors=colors,
        autopct='%1.1f%%',
        explode=(0,0.1,0,0),
        startangle=90)


plt.title("Ticket Priority")
plt.show()
```

```python
plt.figure(figsize=(6,6))
gender = df['Customer Gender'].value_counts()
labels = ['Male','Female','others']
colors = ['red','yellow','green']
plt.pie(gender,
        labels=gender.index,
        colors=colors,
        autopct='%1.1f%%',
        explode=(0,0.1,0),
        startangle=45)
plt.title('gender')
plt.show()
```

```python
plt.figure(figsize=(7,5))

plt.hist(df['Customer Satisfaction Rating'],
         bins=10,color='brown')

plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.title("Customer Satisfaction Distribution")

plt.show()
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

text = " ".join(df['Ticket Description'].astype(str))

wordcloud = WordCloud(
    width=1000,
    height=500,
    background_color='white'
).generate(text)

plt.figure(figsize=(12,6))
plt.imshow(wordcloud)
plt.axis("off")
plt.show()
```

```python
text  = " ".join(df['Customer Name'].astype(str))
wordcloud = WordCloud(
    width = 1000,
    height= 5000,
    background_color='lavender'
).generate(text)
plt.figure(figsize=(20,20))
plt.imshow(wordcloud)
plt.axis('off')
plt.show()
```

```python

plt.figure(figsize=(10,6))#data:Correlation matrix

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,#annot=True	Values 
    cmap='coolwarm',#cmap='coolwarm'	Color theme
    fmt=".2f",#fmt=".2f"	2 decimal places
    linewidths=0.5,#linewidths=0.5	Cell border width
    linecolor='black'#linecolor='black'	Border color
)

plt.title("Correlation Heatmap")
plt.show()
```

```python

plt.figure(figsize=(8,6))

sns.boxplot(
    data=df,
    x='Ticket Priority',
    y='Customer Satisfaction Rating',
    palette='Set2',#color theme
    width=0.5,
    showmeans=True,# shows mwan
    meanprops={#customize mean marker
        "marker":"o",
        "markerfacecolor":"red",
        "markeredgecolor":"black",
        "markersize":8
    }
)

plt.title("Customer Satisfaction by Ticket Priority")
plt.xlabel("Ticket Priority")
plt.ylabel("Customer Satisfaction Rating")

plt.show()
```

1. Red dot > Median → Mean greater than median → High values average ko upar le ja rahi hain.
2. Red dot < Median → Mean less than median → Low values average ko neeche le ja rahi hain.
3. Red dot ≈ Median → Data roughly symmetric (balanced).

1. Median satisfaction rating is 3 for all ticket priorities.
2. Mean and median are nearly equal, indicating a fairly symmetric distribution.
3. No significant outliers are observed.
4. Customer satisfaction does not vary much across ticket priority levels.

"The red dot represents the mean, while the line inside the box represents the median. A slight difference between them indicates a small amount of skewness in the data. In this graph, the difference is very small, which suggests that the customer satisfaction ratings are fairly balanced across all ticket priorities."
