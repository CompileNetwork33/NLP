1. Red dot > Median → Mean greater than median → High values average ko upar le ja rahi hain.
2. Red dot < Median → Mean less than median → Low values average ko neeche le ja rahi hain.
3. Red dot ≈ Median → Data roughly symmetric (balanced).

1. Median satisfaction rating is 3 for all ticket priorities.
2. Mean and median are nearly equal, indicating a fairly symmetric distribution.
3. No significant outliers are observed.
4. Customer satisfaction does not vary much across ticket priority levels.

"The red dot represents the mean, while the line inside the box represents the median. A slight difference between them indicates a small amount of skewness in the data. In this graph, the difference is very small, which suggests that the customer satisfaction ratings are fairly balanced across all ticket priorities."

### Data cleaning

```python
df.isna().sum()
```

```python
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

```python
import re
##remove urls
df["Customer Email"] = df["Customer Email"].str.replace(
    r"http\S+|www\S+",
    "",
    regex=True
)
```

```python
### for emoji 
emoji_pattern = re.compile(
    "["
    "\U0001F600-\U0001F64F"
    "\U0001F300-\U0001F5FF"
    "\U0001F680-\U0001F6FF"
    "\U0001F1E0-\U0001F1FF"
    "]+",
    flags=re.UNICODE
)

df["Ticket Description"] = df["Ticket Description"].apply(
    lambda x: emoji_pattern.sub("", str(x))
)
```

```python
import string

df["Ticket Description"] = df["Ticket Description"].str.translate(
    str.maketrans("", "", string.punctuation)
)
```

```python
df["Ticket Description"] = df["Ticket Description"].str.replace(
    r"\s+",
    " ",
    regex=True
).str.strip()
```
