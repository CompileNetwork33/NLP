```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
from nltk.corpus import stopwords
```

```python
stopword = stopwords.words('english')
stopword
```

```python
corpus=[]
for i in range(0,len(df)):
    review = df['data'][i]
    review = re.sub('[^a-zA-Z]',' ',review)
    review = review.lower()
    review = review.split()
    review = [ps.stem(word) for word in review if  not word in stopword]
    review=' '.join(review)
    corpus.append(review)
```

```python
corpus
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=2500,ngram_range=(1,2))
x = cv.fit_transform(corpus)
```

```python
import numpy as np
np.set_printoptions(edgeitems=40,linewidth=100000,formatter=dict(float=lambda x:"%.3g"% x))
x
```

```python
cv.vocabulary_
```

```python
y = pd.get_dummies(df['labels'])
y = y.iloc[:,0].values
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.naive_bayes import MultinomialNB
labels = MultinomialNB().fit(x_train,y_train)
y_pre = labels.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix,classification_report
print(confusion_matrix(y_test,y_pre))
print(classification_report(y_test,y_pre))
```
